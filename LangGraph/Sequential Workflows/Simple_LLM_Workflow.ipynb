```python
from langgraph.graph import StateGraph, START, END
from langchain_groq import ChatGroq
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
```

```python
model = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
class LLM_State(TypedDict):

  question: str
  answer: str
```

```python
def llm_qa(state: LLM_State) -> LLM_State:

  que = state['question']

  prompt = f'Answer the question {que}'

  ans = model.invoke(prompt).content

  state['answer'] = ans

  return state
```

```python
graph = StateGraph(LLM_State)

graph.add_node('llm_qa', llm_qa)

graph.add_edge(START, 'llm_qa')
graph.add_edge('llm_qa', END)

workflow = graph.compile()
```

```python
in_state = {"question": "how far is moon from earth?"}

output = workflow.invoke(in_state)

print(output)
```
